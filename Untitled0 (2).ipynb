```python

# CELL 3 — Extract discharge measurements from NASA .mat files

import numpy as np
import pandas as pd
from scipy.io import loadmat
from pathlib import Path

records = []

def get_field(obj, name):
    """Read a field from a MATLAB struct or NumPy structured object."""
    if hasattr(obj, name):
        return getattr(obj, name)

    if isinstance(obj, np.void) and obj.dtype.names and name in obj.dtype.names:
        return obj[name]

    if isinstance(obj, np.ndarray) and obj.dtype.names and name in obj.dtype.names:
        return obj[name]

    return None


def unwrap(value):
    """Remove singleton NumPy array wrappers."""
    while isinstance(value, np.ndarray) and value.size == 1:
        value = value.item()
    return value


def as_vector(value):
    """Convert a MATLAB numeric array to a flat float vector."""
    value = unwrap(value)
    if value is None:
        return np.array([], dtype=float)

    try:
        return np.asarray(value, dtype=float).ravel()
    except (TypeError, ValueError):
        return np.array([], dtype=float)


for mat_path in mat_files:
    print("Reading:", Path(mat_path).name)

    mat = loadmat(mat_path, squeeze_me=True, struct_as_record=False)

    # Find the battery variable, e.g. B0005
    battery_keys = [
        key for key in mat.keys()
        if not key.startswith("__")
    ]

    if not battery_keys:
        print("  No battery variable found.")
        continue

    battery_id = battery_keys[0]
    battery = unwrap(mat[battery_id])
    cycles = get_field(battery, "cycle")

    if cycles is None:
        print("  No cycle field found. Available keys:", battery_keys)
        continue

    cycles = np.atleast_1d(cycles)
    battery_records = 0

    for cycle_index, cycle in enumerate(cycles):
        cycle = unwrap(cycle)

        cycle_type = get_field(cycle, "type")
        if cycle_type is None:
            continue

        cycle_type = unwrap(cycle_type)
        if isinstance(cycle_type, bytes):
            cycle_type = cycle_type.decode("utf-8", errors="ignore")

        if str(cycle_type).strip().lower() != "discharge":
            continue

        data = unwrap(get_field(cycle, "data"))
        if data is None:
            continue

        time_s = as_vector(get_field(data, "Time"))
        temp_c = as_vector(get_field(data, "Temperature_measured"))
        voltage_v = as_vector(get_field(data, "Voltage_measured"))
        current_a = as_vector(get_field(data, "Current_measured"))

        ambient = as_vector(get_field(cycle, "ambient_temperature"))
        ambient_c = float(ambient[0]) if ambient.size else np.nan

        # Keep only cycles with the required measurements
        lengths = [
            len(time_s), len(temp_c),
            len(voltage_v), len(current_a)
        ]

        if min(lengths) == 0:
            continue

        n = min(lengths)

        for i in range(n):
            if not np.all(np.isfinite([
                time_s[i], temp_c[i], voltage_v[i], current_a[i]
            ])):
                continue

            records.append({
                "battery_id": battery_id,
                "cycle_id": cycle_index,
                "time_s": float(time_s[i]),
                "temperature_C": float(temp_c[i]),
                "voltage_V": float(voltage_v[i]),
                "current_A": float(current_a[i]),
                "ambient_C": ambient_c
            })

            battery_records += 1

    print("  Discharge measurement rows:", battery_records)


if not records:
    raise ValueError(
        "No discharge records extracted. "
        "Run the diagnostic code below and share its output."
    )

df = pd.DataFrame(records)
df = df.sort_values(
    ["battery_id", "cycle_id", "time_s"]
).reset_index(drop=True)

print("\nExtraction successful!")
print("Total rows:", len(df))
print("Batteries:", df["battery_id"].unique())
print(df.head())
print(df.isna().sum())
```

```python
df = df.sort_values(
    ["battery_id", "cycle_id", "time_s"]
).copy()

group_cols = ["battery_id", "cycle_id"]

# Recent measurement history
for lag in [1, 3, 6]:
    df[f"temp_lag_{lag}"] = (
        df.groupby(group_cols)["temperature_C"].shift(lag)
    )

    df[f"current_lag_{lag}"] = (
        df.groupby(group_cols)["current_A"].shift(lag)
    )

# Target: temperature at or interpolated to t + 60 seconds.
# Do not extrapolate beyond the end of a discharge cycle.
df["target_temp_C"] = np.nan

for _, idx in df.groupby(group_cols, sort=False).groups.items():
    idx = list(idx)

    t = df.loc[idx, "time_s"].to_numpy()
    temp = df.loc[idx, "temperature_C"].to_numpy()

    future_t = t + HORIZON_SECONDS
    valid = (
        np.isfinite(t)
        & np.isfinite(temp)
        & (future_t <= np.max(t))
    )

    target = np.full(len(t), np.nan)

    if valid.sum() > 0:
        target[valid] = np.interp(
            future_t[valid], t, temp
        )

    df.loc[idx, "target_temp_C"] = target

FEATURES = [
    "temperature_C",
    "voltage_V",
    "current_A",
    "ambient_C",
    "time_s",
    "temp_lag_1",
    "temp_lag_3",
    "temp_lag_6",
    "current_lag_1",
    "current_lag_3",
    "current_lag_6"
]

model_df = df.dropna(
    subset=FEATURES + ["target_temp_C", "battery_id"]
).copy()

print("Usable rows:", len(model_df))
print("Battery groups:", model_df["battery_id"].unique())

if model_df["battery_id"].nunique() < 2:
    raise ValueError(
        "At least two battery groups are needed for "
        "the grouped train/test evaluation."
    )
```

```python
X = model_df[FEATURES]
y = model_df["target_temp_C"]
groups = model_df["battery_id"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.25,
    random_state=RANDOM_STATE
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]
y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training batteries:",
      groups.iloc[train_idx].unique())
print("Testing batteries:",
      groups.iloc[test_idx].unique())

model = RandomForestRegressor(
    n_estimators=300,
    max_depth=18,
    min_samples_leaf=2,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"MAE:  {mae:.3f} °C")
print(f"RMSE: {rmse:.3f} °C")
print(f"R²:   {r2:.3f}")
```

```python

# CELL 9 — Interactive ThermoAI temperature prediction

import pandas as pd
import numpy as np

# Use the same FEATURES list and trained model from Cell 5
# This assumes your model variable is named "model".

print("THERMOAI — Battery Temperature Prediction")

current_temp = float(input("Current battery temperature (°C): "))
voltage = float(input("Battery voltage (V): "))
current = float(input("Battery current (A): "))
ambient = float(input("Ambient temperature (°C): "))
elapsed_time = float(input("Elapsed discharge time (seconds): "))

# Set this based on your cell manufacturer's limits
target_temp = float(
    input("Desired battery temperature target (°C): ")
)

# Basic input validation
if not all(np.isfinite(v) for v in [
    current_temp, voltage, current, ambient,
    elapsed_time, target_temp
]):
    raise ValueError("Please enter finite numeric values.")

if voltage <= 0:
    raise ValueError("Voltage must be positive.")

if current_temp <= target_temp:
    print("Battery is currently at or below your chosen target.")

# The original NASA model requires historical lag features.
# Without previous readings, these are approximated using
# the current readings. This reduces prediction reliability.
input_data = {
    "temperature_C": current_temp,
    "voltage_V": voltage,
    "current_A": current,
    "ambient_C": ambient,
    "time_s": elapsed_time,
    "temp_lag_1": current_temp,
    "temp_lag_3": current_temp,
    "temp_lag_6": current_temp,
    "current_lag_1": current,
    "current_lag_3": current,
    "current_lag_6": current
}

X_input = pd.DataFrame([input_data])[FEATURES]

predicted_temp = float(model.predict(X_input)[0])

print("\n----- THERMOAI RESULTS -----")
print(f"Current temperature: {current_temp:.2f} °C")
print(f"Predicted temperature in 60 seconds: {predicted_temp:.2f} °C")
print(f"Chosen target temperature: {target_temp:.2f} °C")

if predicted_temp >= target_temp:
    print("Cooling recommendation: ACTIVATE / INCREASE COOLING")
elif predicted_temp >= target_temp - 2:
    print("Cooling recommendation: MONITOR; cooling may be needed")
else:
    print("Cooling recommendation: NORMAL MONITORING")

print(
    "\nNote: This is a software recommendation, not a validated "
    "pump command or a prediction of coolant temperature."
)
```

```python

# CELL 10 — Accuracy and inference latency

import time
import numpy as np
import pandas as pd
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# 1. Evaluate predictions on the held-out test set
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("THERMOAI MODEL EVALUATION")
print("-" * 35)
print(f"MAE  : {mae:.4f} °C")
print(f"RMSE : {rmse:.4f} °C")
print(f"R²   : {r2:.4f}")

# Optional: percentage of predictions within 1 °C and 2 °C
errors = np.abs(np.asarray(y_test) - np.asarray(y_pred))

print(f"Within ±1 °C : {np.mean(errors <= 1) * 100:.2f}%")
print(f"Within ±2 °C : {np.mean(errors <= 2) * 100:.2f}%")

# 2. Measure single-row model inference latency
# Use the same feature columns used during training.
sample = X_test.iloc[[0]]

# Warm-up: exclude first-call overhead
for _ in range(10):
    model.predict(sample)

# Repeated measurements
latencies_ms = []
for _ in range(1000):
    start = time.perf_counter()
    model.predict(sample)
    end = time.perf_counter()
    latencies_ms.append((end - start) * 1000)

latencies_ms = np.array(latencies_ms)

print("\nINFERENCE LATENCY")
print("-" * 35)
print(f"Median latency : {np.median(latencies_ms):.4f} ms")
print(f"95th percentile: {np.percentile(latencies_ms, 95):.4f} ms")
print(f"Mean latency   : {np.mean(latencies_ms):.4f} ms")
print(f"Maximum latency: {np.max(latencies_ms):.4f} ms")
```
